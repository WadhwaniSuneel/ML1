## Part 0 — Setup and the Data
A hospital wants to flag patients at risk of stroke from routine information. Target: `stroke` (1 = had a stroke, 0 = did not).

**What to do:**
1. Import everything you need (including `imblearn`) and set `SEED = 42`.

```python
# YOUR CODE HERE
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# YOUR CODE HERE
df=pd.read_csv("healthcare-dataset-stroke-data.csv")
df.describe()
df.info()
```

```python
#  Separate features and target
X = df.drop(columns=["stroke"])
y = df["stroke"]

#  Find categorical columns in X
cat_cols = X.select_dtypes(include=["object", "category"]).columns

# 4. Apply categorical encoding
X = pd.get_dummies(
    X,
    columns=cat_cols,
    dtype=int,
    drop_first=True
)
```

```python
# YOUR CODE HERE
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

def evaluate_model(model, X_train, y_train, X_test, y_test):
    # Train the model
    model.fit(X_train, y_train)

    # Predict classes and probabilities
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Calculate the six metrics
    results = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC AUC": roc_auc_score(y_test, y_prob),
        "PR AUC": average_precision_score(y_test, y_prob)
    }

    return results
```

### Task 2.1 — Validation curve
For `max_depth` = 1, 2, …, 15 and `None`, record the train ROC AUC and the 5-fold stratified CV ROC AUC (train data only). Plot both curves on one chart (labels, title, legend) and report the best depth by CV AUC.

```python
# YOUR CODE HERE
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import validation_curve, StratifiedKFold
import numpy as np
import matplotlib.pyplot as plt

# Define tree depths
depths = list(range(1, 16)) + [None]

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=SEED
)

# Create Decision Tree model
dt = DecisionTreeClassifier(random_state=SEED)

# Calculate validation curve
train_scores, cv_scores = validation_curve(
    dt,
    X_train,
    y_train,
    param_name="max_depth",
    param_range=depths,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)

# Calculate mean ROC AUC
train_mean = train_scores.mean(axis=1)
cv_mean = cv_scores.mean(axis=1)

# Find best depth
best_index = np.argmax(cv_mean)

print("Best max_depth:", depths[best_index])
print("Best CV ROC AUC:", round(cv_mean[best_index], 4))

# Plot validation curve
plt.figure(figsize=(10, 6))

x_values = range(1, 17)

plt.plot(x_values, train_mean, marker='o',
         label="Training ROC AUC")

plt.plot(x_values, cv_mean, marker='o',
         label="Cross-Validation ROC AUC")

plt.xlabel("Maximum Tree Depth")
plt.ylabel("ROC AUC")
plt.title("Validation Curve - Decision Tree")
plt.xticks(x_values, [str(d) for d in depths])
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import learning_curve, StratifiedKFold
import numpy as np
import matplotlib.pyplot as plt

# Define training sizes (20% to 100%)
train_sizes = np.linspace(0.2, 1.0, 5)

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=SEED
)

# Create side-by-side plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Compare depth 1 and unlimited depth
depths = [1, None]

titles = [
    "Learning Curve - Depth 1 (Underfitting)",
    "Learning Curve - Unlimited Depth (Overfitting)"
]

for ax, depth, title in zip(axes, depths, titles):

    # Calculate learning curve
    sizes, train_scores, cv_scores = learning_curve(
        DecisionTreeClassifier(
            max_depth=depth,
            random_state=SEED
        ),
        X_train,
        y_train,
        train_sizes=train_sizes,
        cv=cv,
        scoring="roc_auc",
        n_jobs=-1
    )

    # Calculate mean ROC AUC
    train_mean = train_scores.mean(axis=1)
    cv_mean = cv_scores.mean(axis=1)

    # Plot training and CV scores
    ax.plot(
        sizes,
        train_mean,
        marker='o',
        label="Training ROC AUC"
    )

    ax.plot(
        sizes,
        cv_mean,
        marker='o',
        label="Cross-Validation ROC AUC"
    )

    ax.set_xlabel("Number of Training Rows")
    ax.set_ylabel("ROC AUC")
    ax.set_title(title)
    ax.legend()
    ax.grid(True)

plt.tight_layout()
plt.show()
```

### Task 2.2 — Learning curves
Use `learning_curve` (`scoring='roc_auc'`, train sizes 20%→100%) for a tree with `max_depth=1` and a tree with `max_depth=None`. Plot train and CV AUC against the number of training rows, side by side.

```python
# YOUR CODE HERE
import numpy as np
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import roc_auc_score

# Random number generator
rng = np.random.default_rng(SEED)

# Tree depths
depths = [1, 3, None]

# Store final results
results = []

# Repeat experiment for each depth
for depth in depths:

    predictions = []
    auc_scores = []

    # Generate 40 bootstrap samples
    for i in range(40):

        # Random sampling with replacement
        idx = rng.choice(
            len(X_train),
            size=len(X_train),
            replace=True
        )

        X_boot = X_train.iloc[idx]
        y_boot = y_train.iloc[idx]

        # Train Decision Tree
        tree = DecisionTreeClassifier(
            max_depth=depth,
            random_state=SEED
        )

        tree.fit(X_boot, y_boot)

        # Predict stroke probabilities
        probs = tree.predict_proba(X_test)[:, 1]

        predictions.append(probs)

        # Calculate test ROC AUC
        auc = roc_auc_score(y_test, probs)
        auc_scores.append(auc)

    # Convert predictions into NumPy array
    predictions = np.array(predictions)

    # Calculate average prediction variance
    avg_variance = predictions.var(axis=0).mean()

    # Calculate mean AUC of 40 trees
    mean_auc = np.mean(auc_scores)

    # Calculate AUC of averaged predictions
    avg_predictions = predictions.mean(axis=0)

    averaged_auc = roc_auc_score(
        y_test,
        avg_predictions
    )

    # Save results
    results.append({
        "Depth": str(depth),
        "Average Variance": avg_variance,
        "Mean Test AUC": mean_auc,
        "Averaged Prediction AUC": averaged_auc
    })

# Display results table
variance_df = pd.DataFrame(results)

print("Variance Measurement Results:")
display(variance_df.round(4))
```

```python
# YOUR CODE HERE
import matplotlib.pyplot as plt
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score
)


# 1. Train the three models

# Logistic Regression 
lr = LogisticRegression(max_iter=1000, random_state=SEED)

lr.fit(X_train_scaled, y_train)

# Decision Tree 
dt = DecisionTreeClassifier(max_depth=3, random_state=SEED)

dt.fit(X_train, y_train)

# Random Forest 
rf = RandomForestClassifier(n_estimators=200, min_samples_leaf=5, random_state=SEED, n_jobs=-1)

rf.fit(X_train, y_train)


# 2. Get prediction 

lr_probs = lr.predict_proba(X_test_scaled)[:, 1]
dt_probs = dt.predict_proba(X_test)[:, 1]
rf_probs = rf.predict_proba(X_test)[:, 1]

models = {"Logistic Regression": lr_probs, "Decision Tree (Depth 3)": dt_probs, "Random Forest": rf_probs}



# 3. Plot ROC Curves

plt.figure(figsize=(9, 6))

for name, probs in models.items():

    fpr, tpr, thresholds = roc_curve(y_test, probs)

    auc_score = roc_auc_score(y_test, probs)

    plt.plot(fpr, tpr, label=f"{name} (AUC = {auc_score:.3f})")

# Random guessing diagonal
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray",
    label="Random Guessing"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - Model Comparison")
plt.legend()
plt.grid(True)
plt.show()


# 4. Plot Precision-Recall Curves

plt.figure(figsize=(9, 6))

for name, probs in models.items():

    precision, recall, thresholds = precision_recall_curve(y_test, probs)

    ap_score = average_precision_score(y_test, probs)

    plt.plot(recall, precision, label=f"{name} (AP = {ap_score:.3f})")

#  baseline 
stroke_rate = y_test.mean()

plt.axhline(
    y=stroke_rate,
    linestyle="--",
    color="gray",
    label=f"No-Skill Line (Stroke Rate = {stroke_rate:.3f})"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curves - Model Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

### Task 3.2 — Choose a decision threshold
The hospital wants to catch **at least 80%** of strokes.
1. Using **only the training data**, get out-of-fold probabilities for Logistic Regression (`cross_val_predict(..., method='predict_proba')`), then find the highest threshold that gives recall ≥ 0.80 on them.
2. Apply that threshold to the **test** probabilities. Report test recall, precision, the confusion matrix, how many patients are flagged and how many are false alarms.

```python
# YOUR CODE HERE
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.metrics import (
    precision_recall_curve,
    recall_score,
    precision_score,
    confusion_matrix
)

# 1. Define Logistic Regression model

lr_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        max_iter=1000,
        random_state=SEED
    )
)

# 5-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=SEED
)

# 2. Get out-of-fold probabilities

oof_probs = cross_val_predict(
    lr_model,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

# 3. Find highest threshold

precision, recall, thresholds = precision_recall_curve(
    y_train,
    oof_probs
)

target_recall = 0.80

valid_indices = np.where(recall[:-1] >= target_recall)[0]

if len(valid_indices) == 0:
    raise ValueError("No threshold achieves recall >= 0.80")

best_index = valid_indices[-1]
best_threshold = thresholds[best_index]

print("Selected Threshold:", round(best_threshold, 4))
print("OOF Recall:", round(recall[best_index], 4))


# 4. Fit model on full training data


lr_model.fit(X_train, y_train)

test_probs = lr_model.predict_proba(X_test)[:, 1]

# Apply selected threshold
test_predictions = (
    test_probs >= best_threshold
).astype(int)


# 5. Calculate test performance

test_recall = recall_score(
    y_test,
    test_predictions,
    zero_division=0
)

test_precision = precision_score(
    y_test,
    test_predictions,
    zero_division=0
)

cm = confusion_matrix(
    y_test,
    test_predictions,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()

# Number of patients flagged
flagged = np.sum(test_predictions)

# Number of false alarms
false_alarms = fp


# 6. Display results


print("\nTEST RESULTS")

print("Test Recall:", round(test_recall, 4))
print("Test Precision:", round(test_precision, 4))

print("\nConfusion Matrix:")
print(cm)

print("\nTrue Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)

print("\nTotal Patients Flagged:", flagged)
print("Number of False Alarms:", false_alarms)
```

### Task 5.1 — Label noise

Repeat 5 times with different flips and average (1 repeat for fraction 0). Show a table (rows = fraction, columns = models) and a line plot.

```python
# YOUR CODE HERE
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# 1. Function to randomly flip training labels
def flip_labels(y, fraction, random_state):
    rng = np.random.default_rng(random_state)
    noisy_y = np.array(y).copy()

    n_flip = int(len(noisy_y) * fraction)
    flip_indices = rng.choice(
        len(noisy_y), size=n_flip, replace=False
    )

    noisy_y[flip_indices] = 1 - noisy_y[flip_indices]
    return noisy_y


# 2. Define the four models
models = {
    "Logistic Regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=1000, random_state=SEED)
    ),

    "Decision Tree (Depth 3)": DecisionTreeClassifier(
        max_depth=3, random_state=SEED
    ),

    "Decision Tree (No Limit)": DecisionTreeClassifier(
        random_state=SEED
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        min_samples_leaf=5,
        random_state=SEED,
        n_jobs=-1
    )
}


# 3. Test different label noise fractions
fractions = [0, 0.05, 0.10, 0.20]
results = []

for fraction in fractions:

    # One repeat for 0 noise, five repeats otherwise
    repeats = 1 if fraction == 0 else 5

    auc_scores = {name: [] for name in models}

    for repeat in range(repeats):

        # Create noisy training labels
        y_noisy = flip_labels(
            y_train,
            fraction,
            random_state=SEED + repeat
        )

        for name, model in models.items():

            model.fit(X_train, y_noisy)

            # Predict stroke probabilities on untouched test data
            probabilities = model.predict_proba(X_test)[:, 1]

            auc = roc_auc_score(y_test, probabilities)
            auc_scores[name].append(auc)

    # Average ROC AUC across repeats
    row = {
        name: np.mean(scores)
        for name, scores in auc_scores.items()
    }

    results.append(row)


# 4. Display results table
results_df = pd.DataFrame(
    results,
    index=fractions
)

results_df.index.name = "Label Noise Fraction"

print("Average Test ROC AUC:")
display(results_df.round(4))


# 5. Plot the results
plt.figure(figsize=(9, 5))

for column in results_df.columns:
    plt.plot(
        results_df.index,
        results_df[column],
        marker="o",
        label=column
    )

plt.xlabel("Label Noise Fraction")
plt.ylabel("Test ROC AUC")
plt.title("Effect of Label Noise on Model Performance")
plt.xticks(fractions)
plt.legend()
plt.grid(True)
plt.show()
```
